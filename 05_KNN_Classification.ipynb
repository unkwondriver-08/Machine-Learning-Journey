# K-Nearest Neighbors (KNN)

## 1. What is KNN?

K-Nearest Neighbors (KNN) is a **supervised machine learning algorithm** used for classification and regression.

It predicts the output of a new observation by looking at the K closest observations in the training dataset.

- Classification: Assigns the majority class among the K nearest neighbors.
- Regression: Predicts the average (or weighted average) of the K nearest neighbors' values.

KNN is also called a **lazy learning algorithm** because it does not build an explicit model during training. It stores the training data and performs calculations when a prediction is required.

## 2. How Does KNN Work?

1. Select the number of neighbors (K).
2. Calculate the distance between the new observation and all training observations.
3. Identify the K nearest neighbors.
4. For classification, assign the majority class among those neighbors.
5. For regression, calculate their average target value.

## 3. Distance Calculation

KNN commonly uses Euclidean Distance to measure the distance between observations.

\[
d=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
\]

Here, each feature contributes to the distance calculation.

Other distance metrics can also be used, depending on the problem.

## 4. Selecting the Number of Neighbors (K)

A commonly used rule of thumb is the **Square Root Rule**:

\[
K=\sqrt{N}
\]

Where:
- N = Number of observations in the training dataset.
- K = Number of neighbors.

Example: If the training dataset has 400 observations:

\[
K=\sqrt{400}=20
\]

This provides a starting value, not necessarily the optimal K.

The best K can be selected by testing different values using cross-validation.

**Effect of K:**
- Small K: More sensitive to noise; may cause overfitting.
- Large K: Smoother decision boundary; may cause underfitting.

For binary classification, an odd K can help reduce voting ties.

## 5. Feature Scaling

Feature scaling is particularly important in KNN because it is a distance-based algorithm.

If one feature has a much larger numerical range than another, it can dominate the distance calculation.

We commonly use StandardScaler:

\[
z=\frac{x-\mu}{\sigma}
\]

It standardizes features to approximately:
- Mean = 0
- Standard deviation = 1

Example:

```python
from sklearn.preprocessing import StandardScaler

sc_X = StandardScaler()

X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
```

Fit the scaler only on training data and use the same scaler to transform test data.

## 6. Python Implementation

```python
from sklearn.neighbors import KNeighborsClassifier

# Create classifier
classifier = KNeighborsClassifier(
    n_neighbors=11,
    p=2,
    metric='euclidean'
)

# Train model
classifier.fit(X_train, y_train)

# Predict
y_pred = classifier.predict(X_test)
```

### Important Parameters

| Parameter | Meaning |
|---|---|
| `n_neighbors` | Number of nearest neighbors (K) |
| `p=2` | Uses Euclidean distance when the Minkowski metric is selected |
| `metric='euclidean'` | Distance metric used to calculate distances |

## 7. Important Points to Remember

- KNN is a supervised learning algorithm.
- It can perform both classification and regression.
- It is a distance-based algorithm.
- Feature scaling is important.
- It does not require an explicit training phase to learn model parameters.
- Prediction can be computationally expensive for large datasets because distances must be calculated against training observations.
- Choosing an appropriate K is important for model performance.
- KNN can be sensitive to irrelevant features and high-dimensional data.

**Key Takeaway:** KNN predicts an observation based on the nearest observations in the training dataset. The choice of K, distance metric, and feature scaling can significantly affect its performance.

KNN - Predict whether a person will have diabetes or not

````python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix
from sklearn.metrics import f1_score
from sklearn.metrics import accuracy_score
````

````python
dataset = pd.read_csv('diabetes.csv')
print(dataset.head())
print( len(dataset) )
````

````python
zero_not_accepted = ['Glucose', 'BloodPressure', 'SkinThickness', 'BMI', 'Insulin']

for column in zero_not_accepted:
    dataset[column] = dataset[column].replace(0, np.nan)
    mean = int(dataset[column].mean(skipna=True))
    dataset[column] = dataset[column].replace(np.nan, mean)
````

````python
#split dataset
X = dataset.iloc[:, 0:8]
y = dataset.iloc[:, 8]   # [:, 8] - select the 9th column
X_train,X_test, y_train, y_test = train_test_split(X,y, random_state = 0, test_size =0.2)
````

````python
#Feature scaling

sc_X = StandardScaler()
X_train = sc_X.fit_transform(X_train)
X_test = sc_X.transform(X_test)
````

````python
#define the model
classifer = KNeighborsClassifier(n_neighbors =11, p=2, metric = 'euclidean')
````

````python
classifer.fit(X_train, y_train)
````

````python
#predict the result from the model

y_pred = classifer.predict(X_test)
y_pred
````

````python
#evaluate the model
cm = confusion_matrix(y_test, y_pred)
print(cm)
print(f1_score(y_test, y_pred))
````

````python
print(accuracy_score(y_test, y_pred))
````
